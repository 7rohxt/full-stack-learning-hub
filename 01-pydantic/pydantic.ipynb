# Basics

```python
from pydantic import BaseModel # Basemodel has many inbuilt class methods (handles validation, convertion, etc.)
```

```python
class User(BaseModel): # user class inherited from BaseModel
    id: int
    name: str
    is_active: bool = True
```

```python
u = User(id="101", name="Rohit") # "u" is an instance of class "User"
print(u)
```

```python
print(type(u.id)) # --> even when id="101"(str) it got converted to int. But if id="A1" --> returns error
```

```python
print(u.model_dump()) # --> python dict
```

```python
print(u.model_dump_json()) # --> json
```
