# Basics

```python
from pydantic import BaseModel # Basemodel has many inbuilt class methods (handles validation, convertion, etc.)
```

```python
class User(BaseModel): # user class inherited from BaseModel
    id: int
    name: str
    is_active: bool = True
```

```python
u = User(id="101", name="Rohit") # "u" is an instance of class "User"
print(u)
```

```python
print(type(u.id)) # --> even when id="101"(str) it got converted to int. But if id="A1" --> returns error
```

```python
print(u.model_dump()) # --> python dict
```

```python
print(u.model_dump_json()) # --> json
```

# Validation Handling

```python
from pydantic import BaseModel, ValidationError
```

```python
class Product(BaseModel):
    id: int
    price: float
    is_available: bool
```

```python
try:
    p = Product(id="A1", price="twenty five")

except ValidationError as e:
    print(e)
```

# Field

```python
from pydantic import BaseModel, ValidationError, Field
```

```python
class Products(BaseModel):
    id: int = Field(
        ...,
        gt=0,                  # Greater than 0
        description="Product ID"
    )

    price: float = Field(
        ...,
        gt=0,
        le=10000000,
        description="Product name"
    )
    is_available: bool = Field(
        ...
    )
```

```python
try:
    p = Products(id="1", price="-1", is_available="yeah")

except ValidationError as e:
    print(e)
```

# Nested Model

```python
class Address(BaseModel):
    city: str
    pincode: int


class Customer(BaseModel):
    name: str
    address: Address
```

```python
payload = {
    "name": "Rohit",
    "address": {
        "city": "Chennai",
        "pincode": "600001"  
    }
}
```

```python
customer = Customer(**payload)
```

```python
print(customer)
print(type(customer.address))
print(type(customer.address.pincode))
print(customer.model_dump())
```
